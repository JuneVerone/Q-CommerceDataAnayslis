```python
df2 = df.dropna(subset=['City']).copy()
print("Number of rows after dropping rows with missing values in 'City' column:", len(df2))
df2.head()
```

```python
x = df2['Items_Count'].mode()[0] # Find the mode (scalar) of 'Items_Count'
df['Items_Count'] = df['Items_Count'].fillna(x) # Fill missing values with the mode

print("Number of rows after filling missing values in 'Items_Count' column:", len(df2))
print("Items_Count mode:", x)
```
